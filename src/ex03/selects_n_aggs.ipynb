```python
counts = df.groupby("CarNumber")["Fines"].count().sort_values(ascending=False)
counts
```

```python
maxNumberOfFines = counts.iloc[0]  # 4
top_indices = counts[counts == maxNumberOfFines].index
df.loc[df.index.isin(top_indices)]
```

```python
amounts = df.groupby("CarNumber")["Fines"].sum().sort_values(ascending=False)
# amounts = df.groupby("CarNumber")["Fines"].sum().sort_values(ascending=False).head(5)
amounts
```

```python
maxSumOfFines = amounts.iloc[0]  # либо  = amounts.max()
top_indices = amounts[amounts == maxSumOfFines].index
df.loc[df.index.isin(top_indices)]
```
