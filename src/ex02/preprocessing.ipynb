```python
import pandas as pd

df = pd.read_csv("../data/auto.csv", index_col="ID")
df.count()
```

```python
df = df.drop_duplicates(subset=["CarNumber", "Make_n_model", "Fines"], keep="last")
df.count()
```

```python
df.isna().sum()
```

```python
df = df.dropna(axis=1, thresh=len(df) - 500)
df.isna().sum()
```

```python
# df["Refund"] = df["Refund"].fillna(method="ffill")
df["Refund"] = df["Refund"].ffill().bfill()
df.isna().sum()
```

```python
df["Fines"] = df["Fines"].fillna(df["Fines"].mean())
df.isna().sum()
```

```python
df["Make"] = df["Make_n_model"].apply(
    lambda x: x.split()[0] if pd.notna(x) and len(x.split()) > 0 else pd.NA
)
df.head()
```

```python
df["Model"] = df["Make_n_model"].apply(
    lambda x: " ".join(x.split()[1:]) if pd.notna(x) and len(x.split()) > 1 else pd.NA
)
df.head()
```

```python
df = df.drop(columns=["Make_n_model"])
```

```python
df.to_json("../data/auto.json", orient="records", double_precision=1)
```
