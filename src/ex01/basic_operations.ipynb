```python
import pandas as pd

views = pd.read_csv(
    "../data/feed-views.log", header=None, sep="\t", names=["datetime", "user"]
)

views["datetime"] = pd.to_datetime(views["datetime"]).astype("datetime64[ns]")

for attr in ["year", "month", "day", "hour", "minute", "second"]:
    views[attr] = getattr(views["datetime"].dt, attr)
# views.dtypes
# views
```

```python
views["daytime"] = pd.cut(
    views["hour"],
    bins=[0, 4, 7, 11, 17, 20, 24],
    labels=[
        "night",
        "early morning",
        "morning",
        "afternoon",
        "early evening",
        "evening",
    ],
    right=False,
)

views = views.set_index("user")
```
