```python
import pandas as pd
import numpy as np
import requests

df = pd.read_json("../data/auto.json", orient="records")
df["Model"] = df["Model"].replace("", np.nan)
pd.options.display.float_format = "{:.2f}".format
df
```

```python
sample_df = df.sample(n=200, random_state=21, replace=True)
sample_df["Refund"] = df["Refund"].sample(n=200, random_state=21, replace=True).values
sample_df["Fines"] = df["Fines"].sample(n=200, random_state=21, replace=True).values
sample_df
```

```python
concat_rows = pd.concat([df, sample_df], ignore_index=True)
concat_rows.count()
```

```python
np.random.seed(21)
year = pd.Series(np.random.randint(1980, 2020, size=len(concat_rows)), name="Year")
```

```python
fines = pd.concat([concat_rows, year], axis=1)
```

```python
allSurnamesFromFile = pd.read_json("../../datasets/surname.json")
# allSurnamesFromFile
```

```python
surnames_list = allSurnamesFromFile.iloc[1:, 0].tolist()
# surnames_list
```

```python
np.random.seed(21)
unique_cars = fines["CarNumber"].unique()
surname_values = np.random.choice(surnames_list, size=len(unique_cars))
surname_series = pd.Series(surname_values, name="Surname")
surname_series
```

```python
owners = pd.concat([pd.Series(unique_cars, name="CarNumber"), surname_series], axis=1)
owners
```

```python
additional_fines_data = [
    ["A111AA197RUS", 1, 500.0, "Ford", "Focus", 2010],
    ["B222BB197RUS", 2, 1000.0, "Toyota", "Camry", 2015],
    ["C333CC197RUS", 1, 1500.0, "Volkswagen", "Golf", 2012],
    ["D444DD163RUS", 2, 2000.0, "Ford", "Mondeo", 2014],
    ["E555EE178RUS", 1, 2500.0, "Tesla", "Model 3", 2020],
]
new_fines_df = pd.DataFrame(additional_fines_data, columns=fines.columns)
fines = pd.concat([fines, new_fines_df], ignore_index=True)
# fines
```

```python
owners = owners.iloc[:-20]
additional_owners_data = [
    ["X777AA116RUS", "DICAPRIO"],
    ["X888BB116RUS", "PITT"],
    ["X999CC116RUS", "DEPP"],
]
new_owners_df = pd.DataFrame(additional_owners_data, columns=owners.columns)
owners = pd.concat([owners, new_owners_df], ignore_index=True)
owners
```

```python
innerMergedDF = pd.merge(fines, owners, on="CarNumber", how="inner")
innerMergedDF
```

```python
outerMergedDF = pd.merge(fines, owners, on="CarNumber", how="outer")
outerMergedDF
```

```python
leftMergedDF = pd.merge(fines, owners, on="CarNumber", how="left")
leftMergedDF
```

```python
rightMergedDF = pd.merge(fines, owners, on="CarNumber", how="right")
rightMergedDF
```

```python
pd.options.display.float_format = "{:,.2f}".format

fines_pvt = fines.pivot_table(
    index=["Make", "Model"], columns="Year", values="Fines", aggfunc="sum"
)
fines_pvt
```

```python
fines.to_csv("fines.csv", index=False)
owners.to_csv("owners.csv", index=False)
```
